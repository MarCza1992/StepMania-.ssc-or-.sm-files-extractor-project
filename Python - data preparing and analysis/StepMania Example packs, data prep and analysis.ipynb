# Libraries

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Importing Data

## Import Charts Data.csv

```python
Source_Folder = 'Data'
df_ChartsData = pd.read_csv(os.path.join(Source_Folder, 'Charts Data.csv'), sep = ';')

print(df_ChartsData.info())
df_ChartsData.sample(10)
```

## Import Notes Data.csv

```python
df_NotesStats = pd.read_csv(os.path.join(Source_Folder, 'Notes Data.csv'), sep = ';')

print(df_NotesStats.info())
df_NotesStats.sample(10)
```

# Data Preparing

## joining both tables for easier further analysis

```python
df = pd.merge(df_ChartsData, df_NotesStats, on = 'Chart_ID', how = 'inner')
df.info()
df.sample(10)
```

## Data cleaning and formatting

### Checking unique value list in every column

```python
for i in df.columns:
    print(i)
    print(df[i].unique().tolist())
    print('---------------------')
```

### replacing empty values with NaN

```python
for i in df.columns:
    df[i] = df[i].replace([' ', ''], np.nan)
```

### Song_Title column

```python
#fixing Song Title in one song (python automatically converted 'n/a' title to NaN)
df.loc[df['Chart_ID'].between(15581, 15585), 'Song_Title'] = 'n/a'
```

### Song_Length column

```python
#changing format
df['Song_Length'] = pd.to_timedelta(df['Song_Length']).dt.total_seconds().astype(int)
df = df.rename(columns={'Song_Length': 'Song_Length_(seconds)'})
```

```python
#removing rows with very very short Song_Length time, useless for database
df = df[df['Song_Length_(seconds)'] >= 15]
```

### Song_BPM column

```python
#converting Dtype to float
df['MIN_BPM'] = df['MIN_BPM'].str.replace(',', '.').astype(float)
df['MAX_BPM'] = df['MAX_BPM'].str.replace(',', '.').astype(float)
```

### Technical_Notation column

```python
#filling blank data with "Unknown" values
df['Technical_Notation'] = df['Technical_Notation'].fillna('Unknown')
```

### Difficulty_Rating column

```python
#removing rows with absurdly high difficulty rating or 0
df = df[(df['Difficulty_Rating'] <= 25) & (df['Difficulty_Rating'] != 0)]
```

### Type column

```python
#removing rows with "other" values
df = df[df['Type'] != 'other']
```

### Removing rows with nan values in important columns

```python
#removing columns with NaN data
df = df.dropna(subset = ['Difficulty_Name', 'Difficulty_Rating', 'Type', 'Pack', 'Song_Title', 'Song_Length_(seconds)'])
```

### Removing rows with almost empty charts (less than 5 objects)

```python
#creating new column that sum up notes, longnotes and rolls
df['Total_objects_(except_Mines)'] = df['Amount_of_Notes'] + df['Amount_of_LongNotes'] + df['Amount_of_Rolls']
```

```python
#removing almost empty charts
df = df[df['Total_objects_(except_Mines)'] >= 5]
```

### deleting duplicates and reseting indexes

```python
df = df.drop_duplicates(subset = ['Song_Title', 'Song_Subtitle', 'Song_Artist', 'Difficulty_Name', 'Difficulty_Rating', 'Technical_Notation', 'Pack',
                                  'Type', 'Amount_of_Notes', 'Amount_of_LongNotes', 'Amount_of_Rolls', 'Amount_of_Mines'])
```

```python
df.reset_index(drop = True, inplace = True)
```

```python
df.info()
df.sample(15)
```

# Data analysis

```python
# function for print probably in every question
def print_sample(x: str) -> str:
    center = x.center(150)
    text = print(150*'-', '\n', center, '\n', 150*'-')
    return text
```

## 1. What is the distribution of song lengths in terms of seconds?

```python
conditions1 = [
    (df['Song_Length_(seconds)'] < 105),
    (df['Song_Length_(seconds)'].between(105, 150)),
    (df['Song_Length_(seconds)'].between(151, 240)),
    (df['Song_Length_(seconds)'] > 240)]

choices1 = [
    'Short songs (max 00:01:44)',
    'Standard songs (00:01:45 - 00:02:30)',
    'Long songs (00:02:31 - 00:04:00)',
    'Marathon songs (min 00:04:01)']

df_question1 = df
df_question1['Song_Length_Category'] = np.select(conditions1, choices1, default = 'Undefinied')
df_question1 = df_question1.drop_duplicates(subset = ['Song_Title', 'Song_Subtitle', 'Song_Artist'])

sns.histplot(df_question1['Song_Length_Category'], color = 'blue')
plt.xlabel('Category')
plt.xticks(rotation = 90)
plt.ylabel('Song Count')
plt.title('Song distribution in data by Song_Length')
plt.grid(True)
print_sample('the most popular songs are in 00:01:45-00:02:30 interval, which is not surprising, as this is a standardized song length in most official rhythm games. However, when we exclude that interval, custom creators tend to prefer making charts for shorter songs (<00:01:45) rather than longer ones (>00:02:30)')
plt.show()
```

## 2. What is the most often used difficulty rating in every difficulty name?

```python
difficulty_order = {
    'Beginner': 0,
    'Easy': 1,
    'Medium': 2,
    'Hard': 3,
    'Challenge': 4,
    'Edit': 5
}

df_question2 = df.groupby(['Difficulty_Name', 'Difficulty_Rating']).agg(
    Total_of_Charts = ('Chart_ID', 'count')
).reset_index()

df_question2['Difficulty_Name_Order'] = df_question2['Difficulty_Name'].map(difficulty_order)

df_question2 = df_question2.sort_values(['Difficulty_Name_Order', 'Total_of_Charts'], ascending = [True, False]).groupby('Difficulty_Name').head(3)

sns.barplot(data = df_question2, x = 'Difficulty_Rating', y='Total_of_Charts', hue = 'Difficulty_Name')
plt.xlabel('Difficulty Rating')
plt.ylabel('Charts Total')
plt.legend(title = 'Difficulty Name')
plt.title('Top 3 difficulty ratings used for every difficulty')
print_sample('The difficulty distribution is quite clear, with a consistent change in difficulty every 3 ratings')
plt.show()
print_sample('Table view') 
df_question2.drop(columns = 'Difficulty_Name_Order').reset_index(drop = True)
```

## 3. Higher difficulty = higher BPM?

```python
#I will use only top difficulty (Challenge) for that analysis, because thats a hardest difficulty for each songs, lower difficulties MUST be done, so I wont consider it
df_question3 = df[(df['Difficulty_Name'] == 'Challenge') & df['Difficulty_Rating'].between(8, 20)]
df_question3_grouped = df_question3.groupby('Difficulty_Rating').agg(
    Avarage_BPM = ('MAX_BPM', 'mean'))

df_question3_grouped.plot(kind = 'bar')
plt.xlabel('Difficulty Rating (only top difficulties)')
plt.ylabel('Average BPM')
plt.ylim(100,250)
plt.title('Avarage BPM vs Difficulty Rating')
print_sample('Indeed, Step Artists make harder charts when the BPM is higher in key difficulty rating values')
plt.show()
```

## 4. What are the most common Technical Notations used in charts?

```python
#for that question I will use charts with difficulty rating 7 and above, because Technical Notations are rarely used in lower difficulties
df_question4 = df[df['Difficulty_Rating'] >= 7]

Notations = ['BR', 'XO', 'FS', 'SS', 'DS', 'XM', 'RH', 'BU']
Results = []

for i in Notations:
    Number_of_Charts = df_question4[df_question4['Technical_Notation'].str.contains(i, na=False)]['Chart_ID'].count()
    Results.append({'Notation': i, 'Total Charts': Number_of_Charts})

Number_of_Charts = df_question4[~df_question4['Technical_Notation'].str.contains('BR|XO|FS|SS|DS|Unknown', na=False)]['Chart_ID'].count()
Results.append({'Notation': 'Other', 'Total Charts': Number_of_Charts})

df4 = pd.DataFrame(Results)
df4 = df4.sort_values('Total Charts', ascending = False)

sns.barplot(data = df4, x = 'Notation', y = 'Total Charts', hue = 'Notation')
plt.grid(True)
plt.title('Number of charts that use certain notation')
print_sample('The most commonly used techs are BR, XO, and FS. This is not a surprise, as there is even a special shortcut for these three notations - "BXF" - because they are the most frequently used in charts')
plt.show()
```

## 5. Avarage NPS (Notes Per Second) in each difficulty rating

```python
df_question5 = df.copy()
# Notes per second = count/seconds
df_question5['Object_per_Second'] = round(df_question5['Total_objects_(except_Mines)'] / df_question5['Song_Length_(seconds)'], 2)

df_question5_grouped = df_question5.groupby('Difficulty_Rating')['Object_per_Second'].mean().reset_index()

sns.lineplot(data = df_question5_grouped, x = 'Difficulty_Rating', y = 'Object_per_Second', marker = 'o', linestyle = '--', color = 'red')
plt.xlim(1,25)
plt.xticks(range(1,25,1))
plt.xlabel('Difficulty Rating')
plt.ylabel('Average Notes per Second')
plt.title('Average NPS in each difficulty rating')
plt.grid(True)
print_sample('Nothing surprising here, the only anomaly are 22+ ratings but I think its because of too low amount of data in that scenario')
plt.show()
```

## 6. Which packs offer the biggest amount charts which type are not "single"?

```python
# filtering table
df_question6 = df[df['Type'] != 'single']
# grouping
df_question6 = df_question6.groupby(['Type', 'Pack']).agg(
    Total_Charts = ('Chart_ID', 'count'))

df_question6 = df_question6.sort_values(['Type', 'Total_Charts'], ascending = [True, False]).groupby('Type').head(3)
print_sample('We can see that DDR packs offer most of the different gamemodes, while Pump Doubles are practically nonexistent')
df_question6.head(18)
```

## 7. Statistics for top 20 Step Artists (by number of charts)

```python
df_question7 = df.copy()
df_question7['Total_Amount_of_Objects'] = df_question7['Amount_of_Notes'] + df_question7['Amount_of_LongNotes'] + df_question7['Amount_of_Rolls']

df_question7 = df_question7.groupby('Step_Artist').agg(
    Count_of_Charts = ('Chart_ID', 'count'),
    AVG_difficulty = ('Difficulty_Rating', 'mean'),
    AVG_of_objects_placed = ('Total_Amount_of_Objects', 'mean'),
    AVG_Song_Length = ('Song_Length_(seconds)', 'mean')
)

df_question7 = df_question7.sort_values('Count_of_Charts', ascending = False).head(20)
print_sample('Ready statistics for the top 20 charters by the number of charts. We can see that ITGAlex creates the longest, hardest and b charts, while J. Frederick has made the most charts, but most of them are relatively easy (low average difficulty)')
round(df_question7,2)
```

# Custom technical notation table for Power BI

```python
df_custom1 = df

notations = ['XM', 'DS', 'XO', 'FS', 'DT', 'JA', 'RH', 'MA', 'BU', 
            'BR', 'DR', 'JU', 'SJ', 'GH', 'BT', 'SS', 'FL', 'HS', 'CT', 'KS', 'RS']

# scanning every row for notations
for i in notations:
    df_custom1[i] = df_custom1['Technical_Notation'].str.contains(i, na=False).astype(int)

# modifying table with Chart ID and notations only columns
df_custom1 = df_custom1[['Chart_ID'] + notations]

# pivoting and filtering table only for 1 values
df_custom1 = pd.melt(df_custom1, id_vars=['Chart_ID'], var_name='Notation', value_name='Value')
df_custom1 = df_custom1[df_custom1['Value'] == 1]
df_custom1 = df_custom1.sort_values('Chart_ID').reset_index(drop = True)

df_custom1.to_csv('Notation Data.csv', index = False)

df_custom1.head(20)
```
